# Partie 1 : Series

Une **Series** est un tableau 1D étiqueté (un index + des valeurs) : la brique de base de Pandas, équivalente à une colonne de tableau.

```python
# 1) Series de 4 températures
temperatures=pd.Series([24.5, 25.1, 26.3, 27.0])
print(f"1) {temperatures}")

# 2) Index
print(f"2) L'Index: {temperatures.index}\n")

# 3) Première valeur
print(f"3) Première Valeure: {temperatures.iloc[0]}\n")

# 4) Dernière valeur
print(f"4) Dernière valeur: {temperatures.iloc[-1]}\n")
```

```python
# 5) Series indexée par heure (12h, 13h, 14h, 15h)

# Méthode A:
temp_heures_A=pd.Series([24.5, 25.1, 26.3, 27.0],
                       index=["12h", "13h", "14h", "15h"])

temp_heures_B=pd.Series([24.5, 25.1, 26.3, 27.0])
temp_heures_B.index=["12h", "13h", "14h", "15h"]

print(f"5) Méthode A \n {temp_heures_A}\n")
print(f"5) Méthode B \n {temp_heures_B}")
```

```python
# 6) Première et dernière valeur

print("Première valeur (12h) :", temp_heures_A.iloc[0])
print("Dernière valeur (15h) :", temp_heures_B.iloc[-1])
```

# Partie 5 : Manipulation des colonnes

On peut ajouter, renommer ou supprimer des colonnes — une étape courante de préparation des données (*feature engineering*).

```python
# 1) Ajout de la colonne temperature_fahrenheit
df["temperature_fahrenheit"]=df["temperature"] * 9/5+32

# 2) Catégorisation de la température avec np.where()
df["niveau_temperature"]=np.where(df["temperature"] > 30, "Elevé", "Normale")

df[["temperature", "temperature_fahrenheit", "niveau_temperature"]].head()
```

```python
# 3) Renommage de la colonne humidite
df=df.rename(columns={"humidite":"humidite_relative"})

# 4) Suppression de la colonne temperature_fahrenheit
df=df.drop(columns=["temperature_fahrenheit"])

df.columns.tolist()
```

# Partie 6 : Filtrage

Le filtrage permet d'isoler les lignes qui vérifient une ou plusieurs conditions — essentiel pour repérer les situations anormales.

```python
# 1) Températures > 30°C
temp_elevees=df[df["temperature"] > 30 ]
temp_elevees.head()
```

```python
# 2) Température > 30°C ET Humidité > 70
temp_humidite_elevees=df[(df["temperature"] > 30) & (df["humidite_relative"] > 70)]
temp_humidite_elevees.head()
```

# Partie 7 : Tri

Trier un DataFrame permet de faire ressortir les valeurs extrêmes (pics, records) selon une ou plusieurs colonnes.

```python
# 1) Tri croissant par température
df_trie1=df.sort_values("temperature")

# 2) Tri décroissant par température
df_trie2=df.sort_values("temperature", ascending=False)

df_trie1.head()
```

```python
# 3) Top 10 des températures les plus élevées
top10=df.sort_values("temperature", ascending=False).head(10)
top10
```

```python
# 4) Tri par bâtiment (croissant) puis température (décroissant)
df_trie_multi=df.sort_values(["batiment", "temperature"], ascending=[True, False])
df_trie_multi.head(10)
```

# Partie 8 : Analyse

Le regroupement (`groupby`) permet de calculer des statistiques par groupe (ici, par bâtiment) — la base de toute analyse comparative.

```python
# 1) ["consommation"].mean()Consommation moyenne par bâtiment
df.groupby("batiment")["consommation"].mean()
```

```python
# 2) Min, Max, Moyenne, Ecart-type de la température par bâtiment
df.groupby("batiment")["temperature"].agg(["min", "max", "mean", "std"])
```

```python
# 3) Par bâtiment : Max et Moyenne de la température et de la consommation
df.groupby("batiment")[["temperature", "consommation"]].agg(["max", "mean"])
```

```python
# 4) Température moyenne globale
print("Température moyenne globale :", round(df["temperature"].mean(), 2))

# 5) Température maximale
print("Température maximale :", df["temperature"].max())

# 6) Consommation totale 
print("Consommation totale :", round(df["consommation"].sum(), 2))
```

```python
# 7) Bâtiment qui consomme le plus en moyenne
conso_moyenne_batiment= df.groupby("batiment")["consommation"].mean()
print("Bâtiment le plus consommateur :", conso_moyenne_batiment.idxmax(),
     "(", round(conso_moyenne_batiment.max(), 2), ")")
```

```python
# 8) Nombre d'alertes par bâtiment
df[df["etat"] == "ALERTE"].groupby("batiment").size()
```

# Partie 9 : Gestion des valeurs manquantes

Un modèle de Machine Learning ne peut généralement pas être entraîné sur des données incomplètes : il faut détecter puis traiter les valeurs manquantes (`NaN`).

```python
# 2) Nombre total de valeurs manquantes
print("Total de valeurs manquantes :", df.isna().sum().sum())
```

```python
# 3) Taux de valeurs manquantes (%)
taux_manquant=df.isna().sum().sum()/df.size*100
print(f"Taux de valeurs manquantes : {taux_manquant: .2f} %")
```

```python
# 4) Lignes contenant des valeurs manquantes
lignes_manquantes=df[df.isna().any(axis=1)]
lignes_manquantes
```

```python
# 5) Température et humidité : remplacement par la moyenne
df["temperature"]=df["temperature"].fillna(df["temperature"].mean())
df["humidite_relative"]=df["humidite_relative"].fillna(df["humidite_relative"].mean())

# 6) Consommation : remplacement par la mediane
df["consommation"]=df["consommation"].fillna(df["consommation"].median())

# 7) Etat : remplacement par "INCONNU"
df["etat"]=df["etat"].fillna("INCONNU")

# Vérification
df.isna().sum()
```

# Partie 10 : Gestion des doublons

Des mesures dupliquées (ex : double envoi d'un capteur) fausseraient les statistiques et l'entraînement d'un futur modèle : elles doivent être détectées puis supprimées.

```python
# 1) Nombre de doublons
print("Nombre de doublons :", df.duplicated().sum())
```

```python
# 2) Affichage des doublons
df[df.duplicated()]
```

```python
# 3) Suppression des doublons puis vérification
df=df.drop_duplicates()
print("Nombre de doublons après suppresseion :", df.duplicated().sum())
print("Nouvelles dimensions de df :", df.shape)
```

Un dernier tour d'horizon des données nettoyées, avant export — utile pour vérifier la cohérence globale du jeu de données.
