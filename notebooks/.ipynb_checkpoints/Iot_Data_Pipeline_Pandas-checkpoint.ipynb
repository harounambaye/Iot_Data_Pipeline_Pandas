# Partie 1 : Series

Une **Series** est un tableau 1D étiqueté (un index + des valeurs) : la brique de base de Pandas, équivalente à une colonne de tableau.

```python
# 1) Series de 4 températures
temperatures=pd.Series([24.5, 25.1, 26.3, 27.0])
print(f"1) {temperatures}")

# 2) Index
print(f"2) L'Index: {temperatures.index}\n")

# 3) Première valeur
print(f"3) Première Valeure: {temperatures.iloc[0]}\n")

# 4) Dernière valeur
print(f"4) Dernière valeur: {temperatures.iloc[-1]}\n")
```

```python
# 5) Series indexée par heure (12h, 13h, 14h, 15h)

# Méthode A:
temp_heures_A=pd.Series([24.5, 25.1, 26.3, 27.0],
                       index=["12h", "13h", "14h", "15h"])

temp_heures_B=pd.Series([24.5, 25.1, 26.3, 27.0])
temp_heures_B.index=["12h", "13h", "14h", "15h"]

print(f"5) Méthode A \n {temp_heures_A}\n")
print(f"5) Méthode B \n {temp_heures_B}")
```

```python
# 6) Première et dernière valeur

print("Première valeur (12h) :", temp_heures_A.iloc[0])
print("Dernière valeur (15h) :", temp_heures_B.iloc[-1])
```

# Partie 5 : Manipulation des colonnes

On peut ajouter, renommer ou supprimer des colonnes — une étape courante de préparation des données (*feature engineering*).

```python
# 1) Ajout de la colonne temperature_fahrenheit
df["temperature_fahrenheit"]=df["temperature"] * 9/5+32

# 2) Catégorisation de la température avec np.where()
df["niveau_temperature"]=np.where(df["temperature"] > 30, "Elevé", "Normale")

df[["temperature", "temperature_fahrenheit", "niveau_temperature"]].head()
```

```python
# 3) Renommage de la colonne humidite
df=df.rename(columns={"humidite":"humidite_relative"})

# 4) Suppression de la colonne temperature_fahrenheit
df=df.drop(columns=["temperature_fahrenheit"])

df.columns.tolist()
```

# Partie 6 : Filtrage

Le filtrage permet d'isoler les lignes qui vérifient une ou plusieurs conditions — essentiel pour repérer les situations anormales.

```python
# 1) Températures > 30°C
temp_elevees=df[df["temperature"] > 30 ]
temp_elevees.head()
```

```python
# 2) Température > 30°C ET Humidité > 70
temp_humidite_elevees=df[(df["temperature"] > 30) & (df["humidite_relative"] > 70)]
temp_humidite_elevees.head()
```
